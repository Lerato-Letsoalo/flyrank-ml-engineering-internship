## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

The decision I want to improve is which pages should be reviewed first for a possible content refresh or another appropriate action. A content or SEO reviewer would use the ranked recommendations to decide which pages deserve attention before lower-priority pages.

The action could include reviewing a page for a refresh, expansion, protection, pruning, or monitoring, depending on the evidence behind the recommendation. The cost of a wrong recommendation is that limited review time could be spent on a page that does not have a meaningful opportunity, while a more useful page may be missed. A wrong recommendation could also lead to an unnecessary content change. For this reason, the output will be treated as decision-support rather than as an automatic instruction to change a page.
